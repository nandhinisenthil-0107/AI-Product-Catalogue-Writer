```python
import pandas as pd
import gradio as gr
import torch
from transformers import pipeline

print("Libraries imported successfully!")
```

```python
model_name = "openai-community/gpt2"

generator = pipeline(
    "text-generation",
    model=model_name
)

print("AI model loaded successfully!")
```

```python
def generate_catalogue(name, category, price, features, audience):

    prompt = f"""
    Write a professional e-commerce product catalogue.

    Product Name: {name}
    Category: {category}
    Price: {price}
    Features: {features}
    Target Audience: {audience}

    Product Title:
    Product Description:
    Key Features:
    Marketing Highlights:

    Use simple, attractive English.
    Do not invent specifications or guarantees.
    """

    result = generator(
        prompt,
        max_new_tokens=120,
        num_return_sequences=1,
        do_sample=True,
        temperature=0.7,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    return result[0]["generated_text"][len(prompt):].strip()
```

```python
output = generate_catalogue(
    "Wireless Headphones",
    "Electronics",
    "1499 INR",
    "Bluetooth, rechargeable battery, noise reduction",
    "College students"
)

print(output)
```

```python
demo = gr.Interface(
    fn=generate_catalogue,
    inputs=[
        gr.Textbox(label="Product Name"),
        gr.Textbox(label="Product Category"),
        gr.Textbox(label="Price"),
        gr.Textbox(label="Product Features"),
        gr.Textbox(label="Target Audience")
    ],
    outputs=gr.Textbox(
        label="Generated Product Catalogue",
        lines=12
    ),
    title="AI Product and Catalogue Writer",
    description=(
        "Enter product details to generate "
        "catalogue content using AI."
    )
)

demo.launch(share=False)
```

```python
catalogue = pd.DataFrame([{
    "Product Name": "Wireless Headphones",
    "Category": "Electronics",
    "Price": "1499 INR",
    "Features": "Bluetooth, rechargeable battery, noise reduction",
    "Target Audience": "College students",
    "Generated Description": output
}])

catalogue.to_csv(
    "product_catalogue.csv",
    index=False
)

catalogue
```
